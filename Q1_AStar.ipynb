```python
import math
import heapq


graph = {
    'A': {'B': 4, 'C': 3},
    'B': {'A': 4, 'D': 3, 'E': 5},
    'C': {'A': 3, 'E': 4, 'F': 4},
    'D': {'B': 3, 'G': 4, 'H': 7},
    'E': {'B': 5, 'C': 4, 'G': 3},
    'F': {'C': 4, 'G': 4, 'H': 7},
    'G': {'D': 4, 'E': 3, 'F': 4, 'H': 3},
    'H': {'D': 7, 'F': 7, 'G': 3}
}


coordinates = {
    'A': (0, 2), 'B': (2, 4),
    'C': (2, 0), 'D': (4, 4),
    'E': (4, 2), 'F': (4, 0),
    'G': (6, 2), 'H': (8, 2)
}


def heuristic(city,goal):
    x1, y1 = coordinates[city]
    x2, y2 = coordinates[goal]
    
    return math.sqrt((x2-x1)**2+(y2-y1)**2)


def a_star(start,goal):
    open_list = []
    heapq.heappush(open_list,(heuristic(start,goal),0,start,[start]))
    
    best_cost = {start : 0}
    closed = set()
    explored = []
    
    while open_list:
        f, g, city, path = heapq.heappop(open_list)
        
        if city in closed:
            continue
            
        closed.add(city)
        explored.append(city)
            
        if city == goal:
            return path, g, explored
        
        for neighbor, distance in graph[city].items():
            if neighbor in closed:
                continue
            
            new_g = g + distance
            
            if neighbor not in best_cost or new_g < best_cost[neighbor]:
                best_cost[neighbor]= new_g
                new_f = new_g + heuristic(neighbor, goal)
                
                heapq.heappush(open_list,(new_f,new_g,neighbor,path+[neighbor]))
                
    return None, None, explored
                


start = input("Enter the source city from A to H").strip().upper()
goal = input("Enter the destination city from A to H").strip().upper()

if start not in graph or goal not in graph:
    print("Invalid City")
    
else:
    path, cost, explored = a_star(start,goal)
    
    
    if path is None:
        print("No route found")
        
    else:
        print("Shortest route : ","->".join(path))
        print("Total path cost is ",cost,"km")
        print("Number of nodes explored",len(explored))
        print("Exploration order:","->".join(explored))
    



    
```
