```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
```

```python

# Step 1: Load and inspect the dataset
df = pd.read_csv("Student_Performance.csv")

print("Dataset shape:", df.shape)
print("Missing values:\n", df.isnull().sum())
print("First five rows:\n", df.head())
```

```python

# Step 2: Select three numeric academic features
features = ["Hours Studied", "Previous Scores", "Sample Question Papers Practiced"]
X = df[features].copy()
```

```python

# Step 3: Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python

# Step 4: Apply the elbow method
k_values = range(1, 11)
wcss = []

for k in k_values:
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_scaled)
    wcss.append(model.inertia_)

print("\nElbow values (K, WCSS):")

# Plot the elbow graph
plt.figure(figsize=(8, 5))
plt.plot(k_values, wcss, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS / Inertia")
plt.title("Elbow Method - Student Grouping")
plt.xticks(list(k_values))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

# Step 5: Train the final K-Means model
k = 4
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X_scaled)

# Step 6: Examine cluster sizes and averages
print("\nStudents in each cluster:\n",
      df["Cluster"].value_counts().sort_index())

profile = df.groupby("Cluster")[features + ["Performance Index"]].mean().round(2)

print("\nCluster averages:\n", profile)

# Step 7: Apply PCA for two-dimensional visualization
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X_scaled)
centers_2d = pca.transform(kmeans.cluster_centers_)

# Step 8: Plot the student clusters and centroids
plt.figure(figsize=(8, 6))

points = plt.scatter(
    X_2d[:, 0], X_2d[:, 1],
    c=df["Cluster"], cmap="viridis",
    s=9, alpha=0.45
)

plt.scatter(
    centers_2d[:, 0], centers_2d[:, 1],
    c="red", marker="X", s=180, label="Centroids"
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Student Groups from K-Means (2D PCA View)")
plt.colorbar(points, label="Cluster Number", ticks=range(k))
plt.legend()
plt.tight_layout()
plt.savefig("experiment6_q26_student_clusters.png", dpi=180)
plt.show()
```
