```python

# Experiment 6 - Q28: Evaluate K-Means Clustering
# Dataset: Mall_Customers.csv

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA

```

```python
# Step 1: Load the dataset
df = pd.read_csv("Mall_Customers.csv")
print("Dataset shape:", df.shape)
print("Missing values:\n", df.isnull().sum())
print("First five rows:\n", df.head())
```

```python

# Step 2: Select three numerical features and handle missing values
features = ["Age", "Annual Income (k$)", "Spending Score (1-100)"]
X = df[features].copy()
X = X.fillna(X.median())

```

```python
# Step 3: Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 4: Compare K-Means models for K from 1 to 10
k_values = range(1, 11)
wcss = []

```

```python
print("\nK-Means Evaluation:")

for k in k_values:
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X_scaled)
    wcss.append(model.inertia_)

    if k > 1:
        score = silhouette_score(X_scaled, labels)
        print(f"K = {k}: WCSS = {model.inertia_:.2f}, Silhouette = {score:.3f}")
    else:
        print(f"K = {k}: WCSS = {model.inertia_:.2f}")

```

```python
# Step 5: Draw the elbow graph
plt.figure(figsize=(8, 5))
plt.plot(list(k_values), wcss, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS (Inertia)")
plt.title("Elbow Method for Mall Customers")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()

```

```python
# Step 6: Train the final model using K = 6
best_k = 6

kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X_scaled)

```

```python
# Step 7: Display number and characteristics of customers
print("\nNumber of customers in each cluster:")
print(df["Cluster"].value_counts().sort_index())

print("\nAverage feature values for each cluster:")
print(df.groupby("Cluster")[features].mean().round(2))
```

```python

# Step 8: Reduce 3D data to 2D for visualization
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X_scaled)
centers_2d = pca.transform(kmeans.cluster_centers_)

# Step 9: Plot the final clusters
plt.figure(figsize=(8, 6))

points = plt.scatter(
    X_2d[:, 0], X_2d[:, 1],
    c=df["Cluster"], cmap="viridis", s=35
)

plt.scatter(
    centers_2d[:, 0], centers_2d[:, 1],
    c="red", marker="X", s=200, label="Centroids"
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Customer Clusters (K = 6)")
plt.colorbar(points, label="Cluster Number", ticks=range(best_k))
plt.legend()
plt.show()
```
