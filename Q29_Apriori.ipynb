```python

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# Step 1: Load the dataset
df = pd.read_csv("Market_Basket_Optimisation.csv", header=None)

print("Total Transactions:", len(df))

# Step 2: Convert rows into transactions
transactions = []

for i in range(len(df)):
    basket = df.iloc[i].dropna().tolist()
    transactions.append(basket)

# Step 3: One-hot encoding
te = TransactionEncoder()
data = te.fit(transactions).transform(transactions)

basket_df = pd.DataFrame(data, columns=te.columns_)

print("\nOne-Hot Encoded Data:")
print(basket_df.head())

# Step 4: Apply Apriori algorithm
frequent_items = apriori(
    basket_df,
    min_support=0.01,
    use_colnames=True
)

print("\nTotal Frequent Itemsets:", len(frequent_items))
print(frequent_items.head(10))

# Step 5: Generate association rules
rules = association_rules(
    frequent_items,
    num_itemsets=len(df),
    metric="confidence",
    min_threshold=0.30
)

# Step 6: Sort rules by lift
rules = rules.sort_values("lift", ascending=False)

# Step 7: Display support, confidence and lift
result = rules[
    ["antecedents", "consequents", "support", "confidence", "lift"]
].head(10).copy()

result["antecedents"] = result["antecedents"].apply(
    lambda x: ", ".join(sorted(x))
)

result["consequents"] = result["consequents"].apply(
    lambda x: ", ".join(sorted(x))
)

print("\nTotal Association Rules:", len(rules))
print("\nTop 10 Association Rules:")
print(result.to_string(index=False))
```
