```python
import heapq
import math

# Weighted graph of 8 cities
graph = {
    'A': {'B': 4, 'C': 3},
    'B': {'A': 4, 'D': 3, 'E': 5},
    'C': {'A': 3, 'E': 4, 'F': 4},
    'D': {'B': 3, 'G': 4, 'H': 7},
    'E': {'B': 5, 'C': 4, 'G': 3},
    'F': {'C': 4, 'G': 4, 'H': 7},
    'G': {'D': 4, 'E': 3, 'F': 4, 'H': 3},
    'H': {'D': 7, 'F': 7, 'G': 3}
}

# City coordinates
coordinates = {
    'A': (0, 2), 'B': (2, 4),
    'C': (2, 0), 'D': (4, 4),
    'E': (4, 2), 'F': (4, 0),
    'G': (6, 2), 'H': (8, 2)
}

# Heuristic function
def heuristic(city, goal):
    x1, y1 = coordinates[city]
    x2, y2 = coordinates[goal]
    return math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

# Greedy Best-First Search
def greedy_best_first(start, goal):
    open_list = [(heuristic(start, goal), start, [start], 0)]
    visited = set()
    explored = []

    while open_list:
        h, city, path, cost = heapq.heappop(open_list)

        if city in visited:
            continue
        visited.add(city)
        explored.append(city)

        if city == goal:
            return path, cost, explored

        for neighbor, distance in graph[city].items():
            if neighbor not in visited:
                new_path = path + [neighbor]
                new_cost = cost + distance

                heapq.heappush(
                    open_list,
                    (heuristic(neighbor, goal), neighbor, new_path, new_cost)
                )

    return None, None, explored

# User input
print("Available cities:", ", ".join(graph.keys()))

start = input("Enter source city (A-H): ").strip().upper()
goal = input("Enter destination city (A-H): ").strip().upper()

if start not in graph or goal not in graph:
    print("Invalid city. Choose from A to H.")
else:
    path, cost, explored = greedy_best_first(start, goal)

    if path is None:
        print("No route found.")
    else:
        print("Selected route:", " -> ".join(path))
        print("Total path cost:", cost, "km")
        print("Number of nodes explored:", len(explored))
        print("Exploration order:", " -> ".join(explored))
```
