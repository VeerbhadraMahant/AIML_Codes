```python

def is_safe(board, row, col, num):
    for i in range(9):
        if board[row][i] == num:
            return False
        if board[i][col] == num:
            return False
    start_row = (row // 3) * 3
    start_col = (col // 3) * 3
    for r in range(start_row, start_row + 3):
        for c in range(start_col, start_col + 3):
            if board[r][c] == num:
                return False
    return True


def check_board(board):
    for row in range(9):
        for col in range(9):
            num = board[row][col]
            if num != 0:
                board[row][col] = 0
                if not is_safe(board, row, col, num):
                    board[row][col] = num
                    return False
                board[row][col] = num
    return True


def solve(board):
    for row in range(9):
        for col in range(9):
            if board[row][col] == 0:
                for num in range(1, 10):
                    if is_safe(board, row, col, num):
                        board[row][col] = num
                        if solve(board):
                            return True
                        board[row][col] = 0
                return False
    return True


def print_board(board):
    for row in board:
        for num in row:
            print(num, end=" ")
        print()


print("Enter 9 rows with 9 numbers each (0 for empty cells):")
board = []
valid_input = True
for i in range(9):
    try:
        row = list(map(int, input(f"Row {i + 1}: ").split()))
    except ValueError:
        valid_input = False
        break
    if len(row) != 9 or min(row) < 0 or max(row) > 9:
        valid_input = False
        break
    board.append(row)

if not valid_input:
    print("Invalid input. Enter exactly 9 numbers from 0 to 9 in each row.")
elif not check_board(board):
    print("Invalid puzzle: given numbers break Sudoku rules.")
elif solve(board):
    print("\nSolved Sudoku:")
    print_board(board)
    if check_board(board) and all(0 not in row for row in board):
        print("Verification successful: all Sudoku rules are satisfied.")
    else:
        print("Verification failed.")
else:
    print("No solution found. This puzzle is unsolvable.")


# board = [
#     [5, 3, 0, 0, 7, 0, 0, 0, 0],
#     [6, 0, 0, 1, 9, 5, 0, 0, 0],
#     [0, 9, 8, 0, 0, 0, 0, 6, 0],
#     [8, 0, 0, 0, 6, 0, 0, 0, 3],
#     [4, 0, 0, 8, 0, 3, 0, 0, 1],
#     [7, 0, 0, 0, 2, 0, 0, 0, 6],
#     [0, 6, 0, 0, 0, 0, 2, 8, 0],
#     [0, 0, 0, 4, 1, 9, 0, 0, 5],
#     [0, 0, 0, 0, 8, 0, 0, 7, 9]
# ]
```
