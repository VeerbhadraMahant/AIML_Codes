```python

# Experiment 2 - Q6: Examination Timetable using CSP

exams = ["AI", "CN", "DBMS", "OS", "ML", "SE"]
time_slots = ["Slot 1", "Slot 2", "Slot 3", "Slot 4"]

students = {
    "S1": ["AI", "CN", "DBMS", "OS"],
    "S2": ["AI", "ML"],
    "S3": ["CN", "ML"],
    "S4": ["DBMS", "ML"],
    "S5": ["DBMS", "SE"],
    "S6": ["OS", "SE"],
    "S7": ["ML", "SE"],
    "S8": ["AI", "SE"]
}

# Find conflicting examinations
conflicts = []
for i in range(len(exams)):
    for j in range(i + 1, len(exams)):
        for subjects in students.values():
            if exams[i] in subjects and exams[j] in subjects:
                conflicts.append((exams[i], exams[j]))
                break


# Check whether a time slot is safe
def is_safe(exam, slot, schedule, conflicts):
    for exam1, exam2 in conflicts:
        if exam == exam1 and schedule.get(exam2) == slot:
            return False
        if exam == exam2 and schedule.get(exam1) == slot:
            return False
    return True


# Backtracking function
def solve(exams, slots, conflicts, schedule, index):
    if index == len(exams):
        return True
    exam = exams[index]
    for slot in slots:
        if is_safe(exam, slot, schedule, conflicts):
            schedule[exam] = slot
            if solve(exams, slots, conflicts, schedule, index + 1):
                return True
            del schedule[exam]
    return False


# Verify final timetable
def verify(exams, slots, conflicts, schedule):
    if len(schedule) != len(exams):
        return False
    for exam in exams:
        if exam not in schedule or schedule[exam] not in slots:
            return False
    for exam1, exam2 in conflicts:
        if schedule[exam1] == schedule[exam2]:
            return False
    return True


# Display examination and enrollment details
print("Exams:", ", ".join(exams))
print("\nStudent Enrollments:")
for student, subjects in students.items():
    print(student + ":", ", ".join(subjects))

print("\nConflicting Exam Pairs:")
for exam1, exam2 in conflicts:
    print(exam1, "-", exam2)


# Generate timetable using 4 slots
schedule = {}
print("\nTimetable with 4 slots:")

if solve(exams, time_slots, conflicts, schedule, 0):
    for slot in time_slots:
        print(slot + ":")
        for exam in exams:
            if schedule[exam] == slot:
                print(" ", exam)

    if verify(exams, time_slots, conflicts, schedule):
        print("Verification successful: No conflicting exams share a slot.")
    else:
        print("Verification failed.")
else:
    print("No valid timetable. Time slots are insufficient.")


# Check whether 3 slots are sufficient
print("\nChecking with only 3 slots:")
schedule3 = {}

if solve(exams, time_slots[:3], conflicts, schedule3, 0):
    print("A valid timetable exists with 3 slots.")
else:
    print("No valid timetable. 3 slots are insufficient.")
```
