```python

board = [[" ", " ", " "], [" ", " ", " "], [" ", " ", " "]]

def display_board():
    for i in range(3):
        print(" | ".join(board[i]))
        if i < 2:
            print("---------")

def check_winner():
    for i in range(3):
        if board[i][0] == board[i][1] == board[i][2] != " ":
            return board[i][0]
        if board[0][i] == board[1][i] == board[2][i] != " ":
            return board[0][i]
    if board[0][0] == board[1][1] == board[2][2] != " ":
        return board[0][0]
    if board[0][2] == board[1][1] == board[2][0] != " ":
        return board[0][2]
    return None

def is_draw():
    for row in board:
        if " " in row:
            return False
    return check_winner() is None

def minimax(is_computer_turn):
    winner = check_winner()
    if winner == "O":
        return 1
    if winner == "X":
        return -1
    if is_draw():
        return 0

    if is_computer_turn:
        best = -10
        symbol = "O"
    else:
        best = 10
        symbol = "X"

    for row in range(3):
        for col in range(3):
            if board[row][col] == " ":
                board[row][col] = symbol
                score = minimax(not is_computer_turn)
                board[row][col] = " "
                if is_computer_turn:
                    best = max(best, score)
                else:
                    best = min(best, score)
    return best

def computer_move():
    best_score = -10
    best_row = 0
    best_col = 0
    for row in range(3):
        for col in range(3):
            if board[row][col] == " ":
                board[row][col] = "O"
                score = minimax(False)
                board[row][col] = " "
                if score > best_score:
                    best_score = score
                    best_row = row
                    best_col = col
    board[best_row][best_col] = "O"

print("Tic-Tac-Toe: You are X, Computer is O.")
display_board()
while True:
    move = input("Enter row and column (1-3), e.g. 1 2: ").split()
    if len(move) != 2 or not move[0].isdigit() or not move[1].isdigit():
        print("Please enter two numbers.")
        continue
    row = int(move[0]) - 1
    col = int(move[1]) - 1
    if not (0 <= row < 3 and 0 <= col < 3) or board[row][col] != " ":
        print("Invalid move. Try again.")
        continue
    board[row][col] = "X"
    print("Your move:")
    display_board()
    if check_winner() == "X":
        print("You win!")
        break
    if is_draw():
        print("It's a draw!")
        break
    computer_move()
    print("Computer's move:")
    display_board()
    if check_winner() == "O":
        print("Computer wins!")
        break
    if is_draw():
        print("It's a draw!")
        break
```
