```python

from heapq import heappush, heappop

start = ((1, 2, 3), (5, 0, 6), (4, 7, 8))
goal = ((1, 2, 3), (4, 5, 6), (7, 8, 0))

def show_board(state):
    for row in state:
        for number in row:
            if number == 0:
                print("_", end=" ")
            else:
                print(number, end=" ")
        print()
    print()

def manhattan(state):
    distance = 0
    for row in range(3):
        for col in range(3):
            number = state[row][col]
            if number != 0:
                goal_row = (number - 1) // 3
                goal_col = (number - 1) % 3
                distance += abs(row - goal_row) + abs(col - goal_col)
    return distance

def get_neighbors(state):
    for row in range(3):
        for col in range(3):
            if state[row][col] == 0:
                blank_row, blank_col = row, col
    neighbors = []
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        new_row = blank_row + dr
        new_col = blank_col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_board = [list(row) for row in state]
            new_board[blank_row][blank_col], new_board[new_row][new_col] = new_board[new_row][new_col], new_board[blank_row][blank_col]
            new_state = tuple(tuple(row) for row in new_board)
            neighbors.append(new_state)
    return neighbors

def a_star(start, goal):
    open_list = []
    heappush(open_list, (manhattan(start), 0, start))
    visited = set()
    cost = {start: 0}
    parent = {start: None}
    while open_list:
        f, g, current = heappop(open_list)
        if current in visited:
            continue
        visited.add(current)
        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]
            path.reverse()
            return path, len(visited)
        for next_state in get_neighbors(current):
            new_cost = g + 1
            if new_cost < cost.get(next_state, float("inf")):
                cost[next_state] = new_cost
                parent[next_state] = current
                priority = new_cost + manhattan(next_state)
                heappush(open_list, (priority, new_cost, next_state))
    return [], len(visited)

print("Initial state:")
show_board(start)
print("Goal state:")
show_board(goal)
path, explored = a_star(start, goal)
if path:
    for i in range(len(path)):
        print("Step", i)
        show_board(path[i])
    print("Total moves:", len(path) - 1)
    print("States explored:", explored)
else:
    print("No solution found.")
    print("States explored:", explored)
```
